### Streaming and Batch

 Streaming -> It is the process of receiving model output incrementally as it is generated. In LangChain, the stream() method returns an iterator that yields output chunks in real time, allowing applications to display responses progressively instead of waiting for the complete output.

```python
### Using Invoke It will give answer when output completed
model.invoke("write a 50 words of paragrarph about artificial intelligence")
```

```python
### Using Stream It will give answer when output in progress word by word
for chunk in model.stream("write a 50 words of paragrarph about artificial intelligence"):
    print(chunk.text, end=" ", flush=True)
```

### Batch


##### Batch is the process of sending multiple independent requests to a model together so they can be processed efficiently, often in parallel, improving performance and reducing cost.

```python
responses=model.batch(
    ["What is Machine Learning",
     "What is AI Learning",
     "What is Deep Learning"
    ]
)
for response in responses:
    print(response)
```
